## 1) Bruxa de Maria Agnesi

Dada a curva:

$$
y=\frac{1}{1+x^2}=(1+x^2)^{-1}
$$

### a) Derivada

Aplicando a regra da cadeia:

$$
\frac{dy}{dx}=-1(1+x^2)^{-2}\cdot 2x
$$

Logo:

$$
\boxed{\frac{dy}{dx}=\frac{-2x}{(1+x^2)^2}}
$$

### b) Reta tangente no ponto $P\left(-1,\frac12\right)$

O coeficiente angular da reta tangente é a derivada no ponto $x=-1$:

$$
m=y'(-1)=\frac{-2(-1)}{(1+(-1)^2)^2}
$$

$$
m=\frac{2}{(2)^2}=\frac{2}{4}=\frac12
$$

Usando a equação da reta:

$$
y-y_0=m(x-x_0)
$$

Como $P\left(-1,\frac12\right)$:

$$
y-\frac12=\frac12(x+1)
$$

$$
y-\frac12=\frac{x}{2}+\frac12
$$

$$
\boxed{y=\frac{x}{2}+1}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 1000)
y = 1/(1+x**2)
yt = x/2 + 1

plt.figure(figsize=(8,5))
plt.plot(x, y, label='Curva: y = 1/(1+x²)')
plt.plot(x, yt, label='Tangente: y = x/2 + 1')
plt.scatter([-1], [1/2], zorder=5, label='P(-1, 1/2)')
plt.ylim(-1, 3)
plt.grid(True)
plt.legend()
plt.title('Curva e reta tangente')
plt.xlabel('x')
plt.ylabel('y')
plt.show()

x_zoom = np.linspace(-1.2, -0.8, 1000)
y_zoom = 1/(1+x_zoom**2)
yt_zoom = x_zoom/2 + 1

plt.figure(figsize=(8,5))
plt.plot(x_zoom, y_zoom, label='Curva')
plt.plot(x_zoom, yt_zoom, '--', label='Reta tangente')
plt.scatter([-1], [1/2], zorder=5, label='P(-1, 1/2)')
plt.grid(True)
plt.legend()
plt.title('Zoom próximo ao ponto P')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

## 3) Receita total

A quantidade vendida é dada por:

$$
q=f(p)
$$

A receita total é:

$$
R(p)=p\cdot f(p)
$$

### a) Interpretação de $f(20)=10000$ e $f'(20)=-350$

$$
f(20)=10000
$$

Isso significa que, quando o preço é de **R$ 20 por metro**, são vendidos **10000 metros** de tecido.

$$
f'(20)=-350
$$

Isso significa que, próximo de $p=20$, se o preço aumentar em R$ 1, a quantidade vendida diminui aproximadamente **350 metros**.

### b) Encontrar $R'(20)$

Como:

$$
R(p)=p\cdot f(p)
$$

Pela regra do produto:

$$
R'(p)=1\cdot f(p)+p\cdot f'(p)
$$

$$
R'(p)=f(p)+pf'(p)
$$

Substituindo $p=20$:

$$
R'(20)=f(20)+20f'(20)
$$

$$
R'(20)=10000+20(-350)
$$

$$
R'(20)=10000-7000
$$

$$
\boxed{R'(20)=3000}
$$

próximo de $p=20$, aumentar o preço em $R$1  aumenta a receita aproximadamente em R$ 3000.


## 4) Lançamento vertical

A função altura é:

$$
h(t)=2+24,5t-4,9t^2
$$

### a) Velocidade

A velocidade é a derivada da altura:

$$
v(t)=h'(t)
$$

$$
v(t)=24,5-9,8t
$$

$$
\boxed{v(t)=24,5-9,8t}
$$

### b) Velocidade após 2 s e 4 s

Para $t=2$:

$$
v(2)=24,5-9,8(2)=24,5-19,6=4,9
$$

$$
\boxed{v(2)=4,9\;m/s}
$$

Para $t=4$:

$$
v(4)=24,5-9,8(4)=24,5-39,2=-14,7
$$

$$
\boxed{v(4)=-14,7\;m/s}
$$

```python
# Gráfico da velocidade

t = np.linspace(0, 6, 500)
v = 24.5 - 9.8*t

plt.figure(figsize=(8,5))
plt.plot(t, v, label='v(t) = 24,5 - 9,8t')
plt.scatter([2, 4], [4.9, -14.7], zorder=5, label='Pontos t=2 e t=4')
plt.axhline(0, linewidth=1)
plt.grid(True)
plt.legend()
plt.title('Velocidade do projétil')
plt.xlabel('Tempo (s)')
plt.ylabel('Velocidade (m/s)')
plt.show()
```

### c) Altura máxima usando derivadas

A altura máxima ocorre quando a velocidade é zero:

$$
v(t)=0
$$

$$
24,5-9,8t=0
$$

$$
9,8t=24,5
$$

$$
t=\frac{24,5}{9,8}=2,5
$$

Logo, o projétil atinge a altura máxima em:

$$
\boxed{t=2,5\;s}
$$

Agora calculamos a altura:

$$
h(2,5)=2+24,5(2,5)-4,9(2,5)^2

$$

$$

$$
h(2,5)=2+61,25-30,625
$$

$$
\boxed{h(2,5)=32,625\;m}
$$

### d) Gráfico da altura com destaque da altura máxima

```python
# Gráfico da altura

t = np.linspace(0, 5.2, 500)
h = 2 + 24.5*t - 4.9*t**2

t_max = 2.5
h_max = 32.625

plt.figure(figsize=(8,5))
plt.plot(t, h, label='h(t) = 2 + 24,5t - 4,9t²')
plt.scatter([t_max], [h_max], zorder=5, label=f'Altura máxima: ({t_max}, {h_max})')
plt.axhline(0, linewidth=1)
plt.grid(True)
plt.legend()
plt.title('Altura do projétil')
plt.xlabel('Tempo (s)')
plt.ylabel('Altura (m)')
plt.show()
```

## 5) Caixa sem tampa


$$
A = 1200\;cm^2
$$

A caixa tem base quadrada de lado $a$ e altura $h$.

Como a caixa não tem tampa, a área total é:

$$
A=a^2+4ah
$$

Logo:

$$
a^2+4ah=1200
$$

Isolando $h$:

$$
4ah=1200-a^2
$$

$$
h=\frac{1200-a^2}{4a}
$$

### a) Função volume em função de $a$

O volume é:

$$
V=a^2h
$$

Substituindo $h$:

$$
V(a)=a^2\cdot \frac{1200-a^2}{4a}
$$

$$
V(a)=\frac{a(1200-a^2)}{4}
$$

$$
\boxed{V(a)=300a-\frac{a^3}{4}}
$$

### b) Maior volume possível

Derivando:

$$
V'(a)=300-\frac{3a^2}{4}
$$

Para encontrar ponto crítico:

$$
V'(a)=0
$$

$$
300-\frac{3a^2}{4}=0
$$

$$
300=\frac{3a^2}{4}
$$

$$
1200=3a^2
$$

$$
a^2=400
$$

$$
a=20
$$

Como $a$ é medida de comprimento, usamos $a=20\;cm$.

Agora encontramos $h$:

$$
h=\frac{1200-20^2}{4(20)}
$$

$$
h=\frac{1200-400}{80}
$$

$$
h=10\;cm
$$

Volume máximo:

$$
V=20^2\cdot 10
$$

$$
\boxed{V_{max}=4000\;cm^3}
$$

```python
# Gráfico do volume da caixa

a = np.linspace(0.1, np.sqrt(1200), 500)
V = 300*a - (a**3)/4

plt.figure(figsize=(8,5))
plt.plot(a, V, label='V(a) = 300a - a³/4')
plt.scatter([20], [4000], zorder=5, label='Máximo: (20, 4000)')
plt.grid(True)
plt.legend()
plt.title('Volume da caixa sem tampa')
plt.xlabel('a (cm)')
plt.ylabel('Volume (cm³)')
plt.show()
```

## 6) Otimização do tempo de execução

A função é:

$$
T(n)=\frac{400}{n}+0,25n
$$

Reescrevendo:

$$
T(n)=400n^{-1}+0,25n
$$

Derivando:

$$
T'(n)=-400n^{-2}+0,25
$$

$$
T'(n)=-\frac{400}{n^2}+0,25
$$

Para minimizar:

$$
T'(n)=0
$$

$$
-\frac{400}{n^2}+0,25=0
$$

$$
0,25=\frac{400}{n^2}
$$

$$
0,25n^2=400
$$

$$
n^2=1600
$$

$$
n=40
$$

Como $n$ representa quantidade de processadores, usamos $n=40$.

$$
\boxed{n=40\text{ processadores}}
$$

O tempo mínimo será:

$$
T(40)=\frac{400}{40}+0,25(40)
$$

$$
T(40)=10+10=20
$$

$$
\boxed{T_{min}=20\text{ horas}}
$$

## 8) Aceleração, velocidade e distância

Dada a aceleração:

$$
a(t)=t+4
$$

E a velocidade inicial:

$$
v(0)=5
$$

### a) Encontrar a velocidade no instante $t$

Como aceleração é a derivada da velocidade:

$$
a(t)=v'(t)
$$

Então:

$$
v(t)=\int a(t)\,dt
$$

$$
v(t)=\int(t+4)\,dt
$$

$$
v(t)=\frac{t^2}{2}+4t+C
$$

Usando $v(0)=5$:

$$
5=\frac{0^2}{2}+4(0)+C
$$

$$
C=5
$$

Logo:

$$
\boxed{v(t)=\frac{t^2}{2}+4t+5}
$$

### b) Distância percorrida em $0\leq t\leq 10$

Como $v(t)$ é positiva nesse intervalo, a distância é:

$$
S=\int_0^{10} v(t)\,dt
$$

$$
S=\int_0^{10}\left(\frac{t^2}{2}+4t+5\right)dt
$$

$$
S=\left[\frac{t^3}{6}+2t^2+5t\right]_0^{10}
$$

$$
S=\frac{10^3}{6}+2(10)^2+5(10)
$$

$$
S=\frac{1000}{6}+200+50
$$

$$
S=\frac{500}{3}+250
$$

$$
S=\frac{500}{3}+\frac{750}{3}
$$

$$
\boxed{S=\frac{1250}{3}\approx 416,67\;m}
